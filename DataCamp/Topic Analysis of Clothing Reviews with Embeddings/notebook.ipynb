## Load the dataset
Load data and perform basic data checks to ensure you are using relevant data for the analysis

```python
# Load the dataset
import pandas as pd
reviews = pd.read_csv("womens_clothing_e-commerce_reviews.csv")

# Display the first few entries
reviews.head()
```

```python
# Start coding here
# Use as many cells as you need.
# Import the following libraries
import openai
from sklearn.manifold import TSNE
import numpy as np
import matplotlib.pyplot as plt

# Remove any null values in review text
review_texts = reviews["Review Text"].dropna()

# Create and store embeddings for the reviews in one API call
client = openai.OpenAI()
response = client.embeddings.create(
    model="text-embedding-3-small",
    input=review_texts.tolist()
)
response_dict = response.model_dump()
embeddings = [response_data["embedding"] for response_data in response_dict["data"]]

# Apply t-SNE for dimensionality reduction to reduce the embeddings to a 2D numpy array
tsne = TSNE(n_components=2)
embeddings_2d = tsne.fit_transform(np.array(embeddings))

# Plot the results of t-SNE in a 2D visual representation of the reviews
for i, point in enumerate(embeddings_2d):
    plt.scatter(point[0], point[1], alpha=0.5)
plt.title("2D Visualization of Review Embeddings")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
# Import the following libraries
from scipy.spatial import distance
from chromadb.utils.embedding_functions import OpenAIEmbeddingFunction
import os

# Define the topics discussed by some reviews
topics = ["Quality", "Fit", "Style", "Comfort"]

# Create and store embeddings for the topics in one API call
topic_response = client.embeddings.create(
    model="text-embedding-3-small",
    input=topics
)
topic_dict = topic_response.model_dump()
topic_embeddings = [topic_data["embedding"] for topic_data in topic_dict["data"]]

# Function to categorize feedback
def categorize_feedback(text_embedding, category_embeddings):
    distances = [{"distance": distance.cosine(text_embedding, category_embedding), "index": index} for index, category_embedding in enumerate(category_embeddings)]
    index_min = min(distances, key=lambda x: x["index"])
    return topics[index_min["index"]]

# Categorize feedback
feedback_categories = [categorize_feedback(embedding, topic_embeddings) for embedding in embeddings]

# Create a persistant client
client = chromadb.PersistentClient()

# Create a vector database
review_embeddings_database = client.create_collection(
    name="clothing_reviews",
    embedding_function=OpenAIEmbeddingFunction(model_name="text-embedding-3-small", api_key=os.environ["OPENAI_API_KEY"])
)

# Store embeddings into vector database
review_embeddings_database.add(ids=[str(i) for i in range(len(review_texts))], documents=review_texts.tolist())

# Function for similarity search
def find_similar_reviews(texts, db, n=3):
    collection = client.get_collection(
        name="clothing_reviews",
        embedding_function=OpenAIEmbeddingFunction(model_name="text-embedding-3-small", api_key=os.environ["OPENAI_API_KEY"])
    )
    result = collection.query(query_texts=[texts], n_results=n)
    return result

# Apply function to example review and find similar feedback
example_review = "Absolutely wonderful - silky and sexy and comfortable"
most_similar_reviews = find_similar_reviews(example_review, review_embeddings_database, 3)["documents"][0]
print(most_similar_reviews)

# Clean up
client.delete_collection(name="clothing_reviews")
```
